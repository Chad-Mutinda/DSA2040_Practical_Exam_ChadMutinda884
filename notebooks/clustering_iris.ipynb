CLUSTERING.

Used the'Iris' dataset from scikit-learn to:

1. Apply K-Means clustering
2. Experiment, try k=2 and k=4
3. Visualize clusters.
4. Analyze.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler

# Create outputs directory
os.makedirs('visualizations', exist_ok=True)

# Load and preprocess data
iris = load_iris()
X = MinMaxScaler().fit_transform(iris.data)
y = iris.target
feature_names = iris.feature_names

# 1. K-Means with k=3
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X)
ari = adjusted_rand_score(y, clusters)

print(f"Adjusted Rand Index (k=3): {ari:.3f}")

# 2. Elbow Method Experiment
inertias = []
for k in range(1, 6):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(10,5))
plt.plot(range(1,6), inertias, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.xticks(range(1,6))
plt.savefig('visualizations/elbow_curve.png', bbox_inches='tight')
plt.close()

# 3. Cluster Visualization
plt.figure(figsize=(12,5))

# Petal features plot
plt.subplot(121)
scatter = plt.scatter(X[:,2], X[:,3], c=clusters, cmap='viridis')
plt.title('K-Means Clusters (k=3)\nPetal Features')
plt.xlabel('Petal length (normalized)')
plt.ylabel('Petal width (normalized)')
plt.colorbar(scatter, ticks=[0,1,2])

# Comparison with true labels
plt.subplot(122)
true_scatter = plt.scatter(X[:,2], X[:,3], c=y, cmap='viridis')
plt.title('Actual Species Classes\nPetal Features')
plt.xlabel('Petal length (normalized)')
plt.ylabel('Petal width (normalized)')
plt.colorbar(true_scatter, ticks=[0,1,2])

plt.tight_layout()
plt.savefig('visualizations/cluster_comparison.png', bbox_inches='tight')
plt.close()
```
